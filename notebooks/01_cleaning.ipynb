# Data Cleaning
This notebook loads the raw ecommerce transaction dataset, inspects its quality, and applies cleaning steps before downstream analysis.

```python
from pathlib import Path
import sys
import pandas as pd

ROOT = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
sys.path.insert(0, str(ROOT))
from src.rfm import clean_transactions

raw_path = ROOT / 'data' / 'raw' / 'ecommerce_transactions_raw.csv'
df_raw = pd.read_csv(raw_path)
df_raw.head()
```

```python
df_raw.info()
df_raw.isna().sum()
```

```python
df_clean = clean_transactions(df_raw)
df_clean.head()
```

```python
print('Rows before cleaning:', len(df_raw))
print('Rows after cleaning:', len(df_clean))
print('Duplicate Order IDs removed:', df_raw.duplicated(subset=['Order ID']).sum())
print('Missing values after cleaning:')
print(df_clean.isna().sum())
```

```python
clean_path = ROOT / 'data' / 'cleaned' / 'ecommerce_transactions_cleaned.csv'
df_clean.to_csv(clean_path, index=False)
clean_path
```
